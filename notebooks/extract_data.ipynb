```python
import yfinance as yf
import pandas as pd
```

```python
# common function to download the dataset from yahoo finance
def download_dataset(ticker):
    return yf.download(
                ticker,
                period="5y",       # 5 years before today
                end='2026-11-16',  # 'end' is exclusive, so this does not include 2026-11-16
                interval='1d',     # daily
                multi_level_index=False,
                auto_adjust=True,
            )
```

```python
# format column name, add prefix
def format_column_name(df, prefix):
    df.columns = [c if c.lower().startswith(prefix) else f'{prefix}_{c}' for c in df.columns.str.lower()]
```

# 1. Download Tesla stock dataset

```python
# download Tesla stock prices daily for 5 years
df = download_dataset('TSLA')
```

```python
# adding the prefix tsla_ to each column name
format_column_name(df, 'tsla')
df.columns
```

# 5. Combined dataset 

```python
combined_df = pd.concat(
    [d.filter(regex=r'(?i)_close$') for d in [df, df_oil, df_lit, df_ndx]],
    axis=1,
    join='inner',
    sort=True
)
```
